## Feature Engineering

```python

import pandas as pd
import numpy as np

# -----------------------------
# 1. Standardise column names
# -----------------------------
sessions = sessions.rename(columns={"rider_id": "user_id"})

# -----------------------------
# 2. Convert datetime columns
# -----------------------------
riders["signup_date"] = pd.to_datetime(riders["signup_date"], errors="coerce", utc=True)

trips["pickup_time"] = pd.to_datetime(trips["pickup_time"], errors="coerce", utc=True)
trips["dropoff_time"] = pd.to_datetime(trips["dropoff_time"], errors="coerce", utc=True)

drivers["signup_date"] = pd.to_datetime(drivers["signup_date"], errors="coerce", utc=True)
drivers["last_active"] = pd.to_datetime(drivers["last_active"], errors="coerce", utc=True)

sessions["session_time"] = pd.to_datetime(sessions["session_time"], errors="coerce", utc=True)

# Use the latest known activity date as reference point
reference_date = max(
    trips["pickup_time"].max(),
    sessions["session_time"].max(),
    riders["signup_date"].max()
)

# -----------------------------
# 3. Engineer rider features
# -----------------------------
riders["customer_tenure_days"] = (
    reference_date - riders["signup_date"]
).dt.days


#riders["was_referred"] = riders["referred_by"].notna().astype(int)

riders = riders.drop(columns=["referred_by"], errors="ignore")

# -----------------------------
# 4. Engineer trip-level features
# -----------------------------

# Trip duration
trips["trip_duration_minutes"] = (
    trips["dropoff_time"] - trips["pickup_time"]
).dt.total_seconds() / 60

# Haversine distance
def haversine_distance(lat1, lon1, lat2, lon2):
    R = 6371

    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)
    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    )

    c = 2 * np.arcsin(np.sqrt(a))

    return R * c

trips["trip_distance_km"] = haversine_distance(
    trips["pickup_lat"],
    trips["pickup_lng"],
    trips["dropoff_lat"],
    trips["dropoff_lng"]
)

# Monetary features
trips["total_trip_cost"] = (trips["fare"] * trips["surge_multiplier"]) + trips["tip"]

trips["tip_percentage"] = trips["tip"] / trips["fare"].clip(lower=1)

trips["surge_cost"] = trips["fare"] * (trips["surge_multiplier"] - 1)

# Time-based trip behaviour
trips["pickup_hour"] = trips["pickup_time"].dt.hour
trips["is_weekend_trip"] = trips["pickup_time"].dt.dayofweek.isin([5, 6]).astype(int)
trips["is_night_trip"] = trips["pickup_hour"].between(22, 23).astype(int) | trips["pickup_hour"].between(0, 5).astype(int)
trips["is_night_trip"] = trips["is_night_trip"].astype(int)

# -----------------------------
# 5. Merge trips with drivers
# -----------------------------
trip_driver = trips.merge(
    drivers,
    on="driver_id",
    how="left",
    suffixes=("_trip", "_driver")
)

# -----------------------------
# 6. Aggregate trip + driver features per user
# -----------------------------
trip_features = (
    trip_driver.groupby("user_id")
    .agg(
        total_trips=("trip_id", "nunique"),
        total_fare=("fare", "sum"),
        avg_fare=("fare", "mean"),
        total_tip=("tip", "sum"),
        avg_tip=("tip", "mean"),
        avg_surge=("surge_multiplier", "mean"),
        total_trip_cost=("total_trip_cost", "sum"),
        avg_trip_cost=("total_trip_cost", "mean"),
        avg_tip_percentage=("tip_percentage", "mean"),
        total_surge_cost=("surge_cost", "sum"),
        avg_trip_duration_minutes=("trip_duration_minutes", "mean"),
        avg_trip_distance_km=("trip_distance_km", "mean"),
        total_trip_distance_km=("trip_distance_km", "sum"),
        avg_pickup_hour=("pickup_hour", "mean"),
        weekend_trip_ratio=("is_weekend_trip", "mean"),
        night_trip_ratio=("is_night_trip", "mean"),
        unique_drivers=("driver_id", "nunique"),
        avg_driver_rating=("rating", "mean"),
        avg_driver_acceptance_rate=("acceptance_rate", "mean"),
        last_trip_time=("pickup_time", "max"),
        first_trip_time=("pickup_time", "min"),
        most_common_payment_type=("payment_type", lambda x: x.mode().iloc[0] if not x.mode().empty else None),
        most_common_weather=("weather", lambda x: x.mode().iloc[0] if not x.mode().empty else None),
        most_common_vehicle_type=("vehicle_type", lambda x: x.mode().iloc[0] if not x.mode().empty else None),
        trip_city=("city_trip", lambda x: x.mode().iloc[0] if not x.mode().empty else None)
    )
    .reset_index()
)

# Recency features from trips
trip_features["days_since_last_trip"] = (
    reference_date - trip_features["last_trip_time"]
).dt.days

trip_features["customer_trip_lifetime_days"] = (
    trip_features["last_trip_time"] - trip_features["first_trip_time"]
).dt.days.clip(lower=1)

trip_features["trips_per_month"] = (
    trip_features["total_trips"] / (trip_features["customer_trip_lifetime_days"] / 30)
)

# Drop raw date columns after creating useful features
trip_features = trip_features.drop(
    columns=["last_trip_time", "first_trip_time"],
    errors="ignore"
)

# -----------------------------
# 7. Aggregate session features per user
# -----------------------------
session_features = (
    sessions.groupby("user_id")
    .agg(
        total_sessions=("session_id", "nunique"),
        avg_time_on_app=("time_on_app", "mean"),
        total_time_on_app=("time_on_app", "sum"),
        avg_pages_visited=("pages_visited", "mean"),
        total_pages_visited=("pages_visited", "sum"),
        conversion_rate=("converted", "mean"),
        last_session_time=("session_time", "max"),
        first_session_time=("session_time", "min"),
        session_city=("city", lambda x: x.mode().iloc[0] if not x.mode().empty else None)
    )
    .reset_index()
)

session_features["days_since_last_session"] = (
    reference_date - session_features["last_session_time"]
).dt.days

session_features["session_lifetime_days"] = (
    session_features["last_session_time"] - session_features["first_session_time"]
).dt.days.clip(lower=1)

session_features["sessions_per_month"] = (
    session_features["total_sessions"] / (session_features["session_lifetime_days"] / 30)
)

session_features["engagement_score"] = (
    session_features["total_time_on_app"] * session_features["conversion_rate"]
)

session_features = session_features.drop(
    columns=["last_session_time", "first_session_time"],
    errors="ignore"
)

# -----------------------------
# 8. Build final customer-level model dataframe
# -----------------------------
merged = (
    riders
    .merge(trip_features, on="user_id", how="left")
    .merge(session_features, on="user_id", how="left")
)

# -----------------------------
# 9. Create churn target
# -----------------------------
merged["churn"] = (merged["churn_prob"] > 0.5).astype(int)

# Remove leakage column
merged = merged.drop(columns=["churn_prob"], errors="ignore")

# -----------------------------
# 10. Final validation checks
# -----------------------------
print("Model dataframe shape:", merged.shape)
print("Unique users:", merged["user_id"].nunique())
print("Duplicate users:", merged["user_id"].duplicated().sum())
print(merged["churn"].value_counts(normalize=True))

# -----------------------------
# 11. Save final feature-engineered dataset
# -----------------------------
#merged.to_parquet("model_df_customer_level_v2.parquet", index=False)
```

### Feature Selection using Mutual Info Classification

```python
cat_cols = merged.select_dtypes(include='object').columns

for col in cat_cols:
    print(f"{col}: {merged[col].nunique()}")
```

```python
from sklearn.feature_selection import mutual_info_classif

# get our features
# encode categorical
# have our x and y
# fit into mutual info
# get the mi_scores
# assisgn scores to the column names


x = merged.drop(columns=["churn", "churn_prob", "user_id"], errors="ignore")
y = merged["churn"]

date_cols = x.select_dtypes(include=["datetime64[ns]", "datetimetz"]).columns

for col in date_cols:
    x[col + "_year"] = x[col].dt.year
    x[col + "_month"] = x[col].dt.month
    x[col + "_day"] = x[col].dt.day
    x[col + "_dayofweek"] = x[col].dt.dayofweek

# Columns to drop after extracting date features
drop_cols = [
    'trip_id', 'user_id', 'driver_id', 'churn_prob', 'pickup_time', 'dropoff_time',
    'signup_date_rider','session_time','session_id','referred_by','signup_date_driver','last_active'
]

x = x.drop(columns=list(date_cols) + drop_cols, errors='ignore')

x = pd.get_dummies(x, drop_first=True)
x = x.fillna(0)

mi_scores = mutual_info_classif(x, y, random_state=42)

mi_df = pd.DataFrame({
    "feature": x.columns,
    "mi_score": mi_scores
}).sort_values(by="mi_score", ascending=False)

print(mi_df.head(20))
```
